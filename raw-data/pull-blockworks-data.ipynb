```python
import requests
import json
from datetime import datetime
from dotenv import load_dotenv
import os

start_date = '2025-01-01'
end_date = datetime.today().strftime('%Y-%m-%d')

# Load environment variables
load_dotenv("API_KEYS.env")
blockworks_api_key = os.getenv("BLOCKWORKS_API_KEY")
```

```python
# Add directories
parent_root = 'blockworks-raw-data'
if not os.path.exists(parent_root):
        os.makedirs(parent_root)
        print(f'Directory made: {parent_root}')
```

```python
# Function for pulling multiple a metric for one or more projects
def pull_asset_data(asset, metrics, start_date, end_date):
    for metric in metrics:
        # Define the API URL
        url = f'https://api.blockworks.com/v1/metrics/{metric}'
    
        
        headers = {
            "x-api-key": blockworks_api_key
        }
        
        params = {
            'project': asset
        }
        
        # Make the GET request
        response = requests.get(url, headers=headers, params=params)
        
        # Raise an error if the request failed
        response.raise_for_status()
        
        # Parse the response JSON
        data = response.json()
        
        # Save it to a file
        with open(f"blockworks-raw-data/{asset}_{metric}.json", "w") as f:
            json.dump(data, f, indent=2)
        
        print(f"Data saved to blockworks-raw-data/{asset}_{metric}.json")
```

```python
# Call the function
assets = ['solana',
'ethereum'
]

metrics = ['rev-usd'
          ]


for asset in assets:
    print(f'{asset}:')
    pull_asset_data(asset, metrics, start_date, end_date)
    print()

print('Done')
```
