```python
import requests, json

# 1) The “page” you can open freely
landing_url = "https://blockworks.co/analytics/l1"

# 2) The actual data endpoint (slug + date range + interval)
api_url = "https://blockworks.co/api/analytics"
params  = {
    "slug":     "l1",
    "start":    "2023-01-01",
    "end":      "2025-05-30",
    "interval": "day",
}

# 3) Browser-style headers
headers = {
    "User-Agent": (
       "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
       "AppleWebKit/537.36 (KHTML, like Gecko) "
       "Chrome/114.0.0.0 Safari/537.36"
    ),
    "Accept":  "application/json",
    "Referer": landing_url,
}

# 4) Spin up a Session so cookies are preserved
session = requests.Session()
session.headers.update(headers)

# 5) “Warm up” — load the landing page so you pick up any CF or auth cookies
session.get(landing_url).raise_for_status()

# 6) Now hit the API
r = session.get(api_url, params=params)
r.raise_for_status()           # should now be 200 OK
series = r.json()              # likely a dict with a 'series' key, or directly the array

# 7) Persist it
with open("blockworks-raw-data/blockworks_l1_rev-data.json", "w") as f:
    json.dump(series, f, indent=2)

print("✅ saved", 
      len(series.get("series", series)), 
      "records to blockworks_l1_rev-data.json")
```
