```python
import requests
import json
from datetime import datetime
from dotenv import load_dotenv
import os

start_date = '2021-01-01'
end_date = datetime.today().strftime('%Y-%m-%d')

# Load environment variables
load_dotenv("API_KEYS.env")
messari_api_key = os.getenv("MESSARI_API_KEY")
```

```python
def pull_asset_data(asset_name, metrics, start_date, end_date):
    for metric in metrics:
        # Define the API URL
        url = f"https://api.messari.io/metrics/v2/assets/{asset_slug}/metrics/{metric}/time-series/1d?start={start_date}T00%3A00%3A00Z&end={end_date}T00%3A00%3A00Z"
    
        
        headers = {
            "accept": "application/json",
            "x-messari-api-key": messari_api_key
        }
        
        # Make the GET request
        response = requests.get(url, headers=headers)
        
        # Raise an error if the request failed
        response.raise_for_status()
        
        # Parse the response JSON
        data = response.json()
        
        # Save it to a file
        with open(f"messari-raw-data/messari_{asset_name}_{metric}.json", "w") as f:
            json.dump(data, f, indent=2)
        
        print(f"Data saved to messari-raw-data/messari_{asset_name}_{metric}.json")
```

```python
# Call the function
assets = {'hyperliquid': 'hyperliquid',
          'ethereum': 'ethereum',
          'pax-gold': 'pax-gold',
          'bitcoin': 'bitcoin',
          'eigenlayer': 'eigenlayer',
          'usdc': 'usdc'
         }

metrics = ['futures-open-interest',
           'futures-funding-rate',
           'futures-volume',
           'marketcap',
           'price',
           'supply'
          ]


for asset_slug, asset_name in assets.items():
    print(f'{asset_name}:')
    pull_asset_data(asset_name, metrics, start_date, end_date)
    print()

print('Done')
```

```python
# Function to pull data for individual exchanges from messari

def pull_exchange_data(start_date, end_date, exchanges, metrics):
    for exchange in exchanges:
        print(f'{exchange}')
        print()
        for metric in metrics:
            url = f"https://api.messari.io/metrics/v1/networks/{network}/metrics/{metric}/time-series/1d?start={start_date}T00%3A00%3A00Z&end={end_date}T00%3A00%3A00Z"
    
            url = f"https://api.messari.io/metrics/v1/exchanges/{exchange}/metrics/{metric}/time-series/1d?start={start_date}T00%3A00%3A00Z&end={end_date}T00%3A00%3A00Z"
            
            headers = {
                "accept": "application/json",
                "x-messari-api-key": messari_api_key
            }
            
            # Make the GET request
            response = requests.get(url, headers=headers)
            
            # Raise an error if the request failed
            response.raise_for_status()
            
            # Parse the response JSON
            data = response.json()
            
            # Save it to a file
            with open(f"messari-raw-data/messari_{exchange}_{metric}.json", "w") as f:
                json.dump(data, f, indent=2)
            print(f"Data saved to messari-raw-data/messari_{exchange}_{metric}.json")
        print('---------------------------')
        print()
    print('done')
```

```python
# Call the function
start_date = "2024-11-01"
end_date = f"{datetime.today().strftime('%Y-%m-%d')}"
exchanges = ['coinbase', 'gemini', 'kraken']
metrics = ['volume', 'futures-volume']

pull_exchange_data(start_date, end_date, exchanges, metrics)
```
