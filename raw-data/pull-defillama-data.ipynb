```python
# Function to find and return a dict of top N protocols by fees/revenue
def top_n_protocols_by_category(file_path, excluded_slugs, n, categories, metric, sort_order, renaming_dict, force_include):
    with open(file_path, 'r') as f:
        data = json.load(f)

    filtered = []
    for entry in data.get('protocols', []):

        name = entry.get('name')
        slug = entry.get('slug')

        if entry.get('category') != category and name not in force_include.get(category, []):
            continue
        
        if slug in excluded_slugs:
            continue

        value = entry.get(sort_order)
        if value is None:
            continue

        for defillama_name, cleaned_name in renaming_dict.items():
            if entry.get('name') == defillama_name:
                name = cleaned_name[0]
                slug = cleaned_name[1]

        filtered.append((slug, name, value))

    sorted_filtered = sorted(filtered, key=lambda x: x[2], reverse=True)
    top_dict = {slug: name for slug, name, _ in sorted_filtered[:n]}
    return top_dict
```

```python
import requests
import json

# Define a category then pull time series fees data for it

def pull_category_defillama(file_path, category):
    with open (file_path, 'r') as f:
        data = json.load(f)

    # drill into the protocols list
    protocols = data.get('protocols',[])
    
    # extract names of entries in the chosen category
    names = [
    entry['name']
    for entry in protocols
    if entry.get('category') == f'{category}'
    ]

    return names

# Pull category names

file_path = 'defillama-raw-data/defillama_fees_all-protocols.json'
category = 'Telegram Bot'
breakout_asset = 'Axiom'
names = pull_category_defillama(file_path = file_path, category = category)

def pull_fees_data(asset):
    # Define the API URL
    metrics = {'dailyFees': 'Fees'
              }
    for metric_slug, metric_name in metrics.items():
        url = f"https://api.llama.fi/summary/fees/{asset}?dataType={metric_slug}"
        
        # Make the GET request
        response = requests.get(url)
        
        # Raise an error if the request failed
        response.raise_for_status()
        
        # Parse the response JSON
        data = response.json()
        
        # Save it to a file
        with open(f"defillama-raw-data/{metric_slug}_data/{asset}_{metric_name}.json", "w") as f:
            json.dump(data, f, indent=2)
        
        print(f"Data saved to defillama-raw-data/{metric_slug}_data/{asset}_{metric_name}.json")


# Call the function
for asset in names:
    print(f'{asset}:')
    pull_fees_data(asset)
    print()

print(f'{breakout_asset}')
pull_fees_data(breakout_asset)
print()
print('Done')
```

```python
# Function to find and return a dict of top N stablecoins as {slug: name}
def top_n_stablecoins_by_mcap(n, pegMechanism, excluded_assets):
    with open('defillama-raw-data/all-stablecoins.json', 'r') as f:
        data = json.load(f)

    assets = data['peggedAssets']
    
    filtered = []
    for entry in assets:
        try:
            if entry["name"] in excluded_assets:
                continue
            if entry['pegMechanism'] != pegMechanism:
                continue
            mcap = entry['circulating']['peggedUSD']
            filtered.append((entry["symbol"], entry["name"], mcap))
        except KeyError:
            continue

    sorted_filtered = sorted(filtered, key=lambda x: x[2], reverse=True)
    top_dict = {symbol: name for symbol, name, _ in sorted_filtered[:n]}
    top_names = list(top_dict.values())
    
    # Save to disk
    with open(f'defillama-raw-data/top-{pegMechanism}-stablecoins-by-mcap.json', 'w') as f:
        json.dump(top_names, f, indent=2)
    
    print(f"Data saved to defillama-raw-data/top-{pegMechanism}-stablecoins-by-mcap.json")
    
    return top_names, top_dict
```
