```python
# Pull TVL for HyperEVM protocols

import requests
import json
import time

# define top HyperEVM protocols tracked by DefiLlama
hyperevm = [
    'felix',
    'hyperlend',
    'hypurrfi',
    'hyperswap',
    'kittenswap-finance',
    'looped-hype',
    'sentiment',
    'laminar',
    'upshift',
    'keiko-finance',
    'valantis',
    'hyperyield',
    'magpie-ecosystem',
    'growihf',
    'timeswap',
    'fan.fun'
]

# Set up hyperevm loop
for i in hyperevm:
    # Define the API URL
    url = f"https://api.llama.fi/protocol/{i}"
    
    # Make the GET request
    response = requests.get(url)
    
    # Raise an error if the request failed
    response.raise_for_status()
    
    # Parse the response JSON
    data = response.json()
    
    # Save it to a file
    with open(f"defillama-raw-data/hyperevm/defillama_{i}_tvl.json", "w") as f:
        json.dump(data, f, indent=2)
    
    print(f"Data saved to defillama-raw-data/hyperevm/defillama_{i}_tvl.json")
    time.sleep(1)

print("Done")
```

```python
# Function to find and return a dict of top N protocols by fees/revenue
def top_n_protocols_by_category(file_path, excluded_slugs, n, category, metric, sort_order):
    with open(file_path, 'r') as f:
        data = json.load(f)

    filtered = []
    protocols = data['protocols']
    for entry in data.get('protocols', []):
        try:
            if entry['category'] != category:
                continue
            slug = entry.get('slug')
            if slug in excluded_slugs:
                continue
    
            metric_value = entry.get(sort_order)
            if metric_value is None:
                continue
    
            name = entry.get('name')
            if not name:
                continue
    
            filtered.append((slug, name, metric_value))
        except:
            continue

    sorted_filtered = sorted(filtered, key=lambda x: x[2], reverse=True)
    top_dict = {slug: name for slug, name, _ in sorted_filtered[:n]}
    return top_dict

```

```python
import requests
import json

categories = {
    'perps_dexs':{'hyperliquid': 'hyperliquid',
          'drift': 'drift',
          'dydx': 'dydx',
          'jupiter': 'jupiter',
          'gmx': 'gmx',
          'apex-pro': 'apex'
             },
    'smart_contract_chains': {'ethereum': 'ethereum',
                         'solana': 'solana',
                         'bsc': 'bnb',
                         'avalanche': 'avalanche',
                         'arbitrum': 'arbitrum',
                         'op mainnet': 'optimism',
                         'hyperliquid': 'hyperliquid',
                         'ton': 'ton',
                         'polygon': 'polygon',
                         'tron': 'tron'
             },
    'borrow_lend_protocols' : {'aave': 'aave',
                         'morpho': 'morpho',
                         'compound-finance': 'compound',
                         'kamino': 'kamino'
                        },
    'launchpads' : {'pump.fun': 'pumpfun',
              'uniderp': 'uniderp',
              'launch coin on believe': 'launch-coin-on-believe',
              'launchlab': 'launchlab'
             }
}

def pull_fees_and_revenue_data(asset_slug):
    # Define the API URL
    metrics = {'dailyRevenue': 'Revenue',
               'dailyFees': 'Fees'
              }
    for metric_slug, metric_name in metrics.items():
        url = f"https://api.llama.fi/summary/fees/{asset_slug}?dataType={metric_slug}"
        
        # Make the GET request
        response = requests.get(url)
        
        # Raise an error if the request failed
        response.raise_for_status()
        
        # Parse the response JSON
        data = response.json()
        
        # Save it to a file
        with open(f"defillama-raw-data/defillama_{asset_name}_{metric_name}.json", "w") as f:
            json.dump(data, f, indent=2)
        
        print(f"Data saved to defillama-raw-data/defillama_{asset_name}_{metric_name}.json")


# Call the function
for category, assets in categories.items():
    print(f'{category}:') 
    print()
    for asset_slug, asset_name in assets.items():
        print(f'{asset_name}:')
        pull_fees_and_revenue_data(asset_slug)
        print()
    print('---------------------------')
print('Done')
```

```python
import requests
import json

# Define a category then pull time series fees data for it

def pull_category_defillama(file_path, category):
    with open (file_path, 'r') as f:
        data = json.load(f)

    # drill into the protocols list
    protocols = data.get('protocols',[])
    
    # extract names of entries in the chosen category
    names = [
    entry['name']
    for entry in protocols
    if entry.get('category') == f'{category}'
    ]

    return names

# Pull category names

file_path = 'defillama-raw-data/defillama_fees_all-protocols.json'
category = 'Telegram Bot'
breakout_asset = 'Axiom'
names = pull_category_defillama(file_path = file_path, category = category)

def pull_fees_data(asset):
    # Define the API URL
    metrics = {'dailyFees': 'Fees'
              }
    for metric_slug, metric_name in metrics.items():
        url = f"https://api.llama.fi/summary/fees/{asset}?dataType={metric_slug}"
        
        # Make the GET request
        response = requests.get(url)
        
        # Raise an error if the request failed
        response.raise_for_status()
        
        # Parse the response JSON
        data = response.json()
        
        # Save it to a file
        with open(f"defillama-raw-data/defillama_{asset}_{metric_name}.json", "w") as f:
            json.dump(data, f, indent=2)
        
        print(f"Data saved to defillama-raw-data/defillama_{asset}_{metric_name}.json")


# Call the function
for asset in names:
    print(f'{asset}:')
    pull_fees_data(asset)
    print()

print(f'{breakout_asset}')
pull_fees_data(breakout_asset)
print()
print('Done')
```
